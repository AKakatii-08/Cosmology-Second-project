```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a=1
b=2
N=10
x=np.linspace(0,a,N++1)
y=np.linspace(0,b,N+1)
print(x)
print(y)
X,Y=np.meshgrid(x,y)
print(X)
print(Y)
```

```python
f= np.sin(X)*np.cos(Y)
#plt.contour(X,Y,f) it is simply contour line
plt.contourf(X,Y,f, levels=20) #it fills the gap
plt.colorbar(label="f") # Display the color bar with the value
plt.xlabel("x")
plt.ylabel("y")
```
