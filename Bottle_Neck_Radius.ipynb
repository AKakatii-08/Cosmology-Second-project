```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Physical Constants
# =========================================================

G = 6.67430e-11          # m^3 kg^-1 s^-2 (SI unit)
c = 3e8         # m/s (SI unit)
M_sun = 2e30       # kg (SI unit)
pc = 3.086e16            # meters (SI unit, 1 parsec in meters)
rho1 = 10**3 * M_sun / pc**3 # kg/m^3 (SI unit)
sigma1 = 300 * 1000 # m/s (SI unit, converted from 300 km/s)
rho2 = 10**6 * M_sun / pc**3 # kg/m^3 (SI unit)
sigma2 = 100 * 1000 # m/s (SI unit, converted from 300 km/s)
H = 14.3 # Unit unclear. Please specify the unit of H to ensure it's in SI.
```

```python
M = np.logspace(3, 8, 100)
# Convert mass to kg
Mkg = M * M_sun
#Bottle neck radius
a1=((256*sigma1*G**2*Mkg**3*2)/(5*c**5*H*rho1))**(1/5)
a2=((256*sigma2*G**2*Mkg**3*2)/(5*c**5*H*rho2))**(1/5)
#print(f"a in meters: {a}")
#print(f"a in meters: {a}")

a_pc1 = a1 / pc
a_pc2 = a2 / pc
#print(f"a in parsecs: {a_pc}")
```

```python
plt.figure(figsize=(10, 6))
plt.loglog(M, a_pc1, linestyle='-',color='goldenrod', label='Large Galaxy')
plt.loglog(M, a_pc2, linestyle='-',color='darkslateblue', label='Small High Density Galaxy')

# Remove minor ticks
plt.tick_params(axis='both', which='minor', bottom=False, left=False)

plt.xlabel('Mass (M_sun)')
plt.ylabel('a (parsec)')
#plt.title('Bottleneck Radius (a) vs. Mass (M)')
#plt.grid(True, which="major", ls="-")
plt.legend()
plt.savefig('bottleneck_radius.png', dpi=600)
plt.show()
```
